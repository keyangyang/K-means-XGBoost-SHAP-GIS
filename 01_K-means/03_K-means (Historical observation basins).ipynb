```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
```

```python
#Extract variable
datA = pd.read_csv(r'02_K-means_his.csv')
data1 = datA['ELE']      #elevation
data1=np.array(data1)
data2 = datA['SLP']      #slope
data2=np.array(data2)
data3 = datA['TMP']      #temperature
data3=np.array(data3)
data4 = datA['PPT']      #precipitation
data4=np.array(data4)
data5 = datA['AREA']     #Basin area
data5=np.array(data5)

# 0-1 Normalization: Normalize by column
data1_normalized = (data1 - np.min(data1)) / (np.max(data1) - np.min(data1))
data2_normalized = (data2 - np.min(data2)) / (np.max(data2) - np.min(data2))
data3_normalized = (data3 - np.min(data3)) / (np.max(data3) - np.min(data3))
data4_normalized = (data4 - np.min(data4)) / (np.max(data4) - np.min(data4))
data5_normalized = (data5 - np.min(data5)) / (np.max(data5) - np.min(data5))

DATA= np.vstack((data1_normalized,data2_normalized,data3_normalized,data4_normalized,data5_normalized)).T
print(DATA)
```

```python
#Perform multi-dimensional k-means clustering and output the cluster centers
julei = KMeans(n_clusters=12, random_state=42) 
julei.fit(DATA)

label = julei.labels_
print("Cluster label:")
print(label)
datA['Cluster'] = label

cluster_centers = julei.cluster_centers_
print("\nCluster center:")
print(cluster_centers)

cluster_centers_output_path = r'06_K12cluster_centers.txt'
with open(cluster_centers_output_path, 'w') as f:
    for center in cluster_centers:
        formatted_center = np.array_str(center)[1:-1] 
        f.write(f"[{formatted_center}],\n") 
print(f"Cluster centers have been saved to the file:{cluster_centers_output_path}")


output_path = r'07_12K_his.csv'
datA.to_csv(output_path, index=False)
print(f"Cluster labels have been saved to the file:{output_path}")
```
