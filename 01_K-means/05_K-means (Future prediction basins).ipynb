```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
```

```python
# Use the given cluster centers to cluster the global flows that need to be predicted
# The given cluster centers are as follows:
init_centers = np.array([
    [0.1434194, 0.59388503, 0.32230225, 0.07028912, 0.02479018],
    [0.15078661, 0.56684446, 0.89257778, 0.03505447, 0.01212893],
    [0.12452018, 0.60538852, 0.91312265, 0.58987749, 0.01604177],
    [0.82006785, 0.70775165, 0.50097417, 0.08077729, 0.00847637],
    [0.12635583, 0.58096882, 0.68690305, 0.13368898, 0.0135437 ],
    [0.34049674, 0.64744453, 0.59932711, 0.08522476, 0.01121737],
    [0.10485024, 0.56119213, 0.87728555, 0.29807345, 0.51071804],
    [0.30446539, 0.61234661, 0.82053632, 0.14782458, 0.01108607],
    [0.45417205, 0.58277838, 0.16279864, 0.17463462, 0.00627457],
    [0.14833351, 0.5910414,  0.86849806, 0.33987251, 0.01964907],
    [0.12945053, 0.56912597, 0.9213201, 0.19367179, 0.02417077],
    [0.1311968, 0.57741507, 0.49841305, 0.1355542, 0.03186765]
])
```

```python
#Data Reading and Processing
data = pd.read_csv('04_K-means_future.csv') 

data1 = data['ELE']     
data1=np.array(data1)
data2 = data['SLP']   
data2=np.array(data2)
data3 = data['TMP']   
data3=np.array(data3)
data4 = data['PPT']   
data4=np.array(data4)
data5 = data['AREA']   
data5=np.array(data5)

data1_normalized = (data1 - np.min(data1)) / (np.max(data1) - np.min(data1))
data2_normalized = (data2 - np.min(data2)) / (np.max(data2) - np.min(data2))
data3_normalized = (data3 - np.min(data3)) / (np.max(data3) - np.min(data3))
data4_normalized = (data4 - np.min(data4)) / (np.max(data4) - np.min(data4))
data5_normalized = (data5 - np.min(data5)) / (np.max(data5) - np.min(data5))

DATA= np.vstack((data1_normalized,data2_normalized,data3_normalized,data4_normalized,data5_normalized)).T#转为矩阵
print(DATA)
```

```python
# Perform KMeans clustering using specified cluster centers
julei = KMeans(n_clusters=12, init=init_centers, n_init=1, random_state=42) 
julei.fit(DATA)  

label = julei.labels_

data['Cluster'] = label

output_path = r'08_12K_future.csv'
data.to_csv(output_path, index=False)
print(f"Cluster labels have been saved to the file:{output_path}")
```
