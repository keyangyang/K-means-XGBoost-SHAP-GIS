```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
```

```python
#提取变量
datA = pd.read_csv(r'02_K-means_his.csv')
data1 = datA['ELE']      #海拔
data1=np.array(data1)
data2 = datA['SLP']      #坡度
data2=np.array(data2)
data3 = datA['TMP']      #气温
data3=np.array(data3)
data4 = datA['PPT']      #降水
data4=np.array(data4)
data5 = datA['AREA']     #流域面积
data5=np.array(data5)

# 0-1标准化：按列进行标准化
data1_normalized = (data1 - np.min(data1)) / (np.max(data1) - np.min(data1))
data2_normalized = (data2 - np.min(data2)) / (np.max(data2) - np.min(data2))
data3_normalized = (data3 - np.min(data3)) / (np.max(data3) - np.min(data3))
data4_normalized = (data4 - np.min(data4)) / (np.max(data4) - np.min(data4))
data5_normalized = (data5 - np.min(data5)) / (np.max(data5) - np.min(data5))

DATA= np.vstack((data1_normalized,data2_normalized,data3_normalized,data4_normalized,data5_normalized)).T#转为矩阵
print(DATA)
```

```python
#进行多维kmeans聚类，输出聚类中心
julei = KMeans(n_clusters=12, random_state=42) #输入聚类数，如n_clusters=4，分为4个聚类（0、1、2、3）
julei.fit(DATA)#对聚类的数据进行聚类

# 获取聚类标签
label = julei.labels_
print("聚类标签：")
print(label)
# 将聚类标签添加到原始数据的最后一列
datA['Cluster'] = label

# 获取每个聚类的中心（即聚类规则）
cluster_centers = julei.cluster_centers_
print("\n聚类中心：")
print(cluster_centers)

# 设置文件保存路径
cluster_centers_output_path = r'06_K12cluster_centers.txt'
# 打开文件并逐行写入聚类中心
with open(cluster_centers_output_path, 'w') as f:
    for center in cluster_centers:
        # 格式化每个聚类中心
        formatted_center = np.array_str(center)[1:-1]  # 去掉方括号
        f.write(f"[{formatted_center}],\n")  # 加上方括号并以逗号结尾
print(f"聚类中心已保存到文件：{cluster_centers_output_path}")

# 保存为新的 CSV 文件
output_path = r'07_12K_his.csv'
datA.to_csv(output_path, index=False)
print(f"聚类标签已保存到文件：{output_path}")
```
